## Question 1
### What is a null hypothesis (H₀) and why is it important in hypothesis testing?

The **null hypothesis (H₀)** is a statement that there is **no significant effect, difference, or relationship** in the population being studied.

It is important because hypothesis testing starts by assuming that H₀ is true. We then use sample data to determine whether there is enough statistical evidence to **reject H₀** in favor of the alternative hypothesis (H₁).

**Example:** If a company claims that the average complaint-resolution time is 10 minutes:

- H₀: μ = 10 minutes
- H₁: μ ≠ 10 minutes

## Question 2
### What does the significance level (α) represent in hypothesis testing?

The **significance level (α)** is the maximum probability of making a **Type I error**, which means rejecting the null hypothesis when it is actually true.

For example, if **α = 0.05**, we accept a 5% risk of rejecting a true null hypothesis.

A common decision rule is:

- If **p-value ≤ α** → Reject H₀
- If **p-value > α** → Fail to reject H₀

## Question 3
### Differentiate between Type I and Type II errors.

| Error | Meaning | Symbol |
|---|---|---|
| **Type I error** | Rejecting H₀ when H₀ is actually true | α |
| **Type II error** | Failing to reject H₀ when H₀ is actually false | β |

**Example:** Suppose H₀ says that a new training program has no effect.

- **Type I error:** We conclude that the training program improves productivity when it actually does not.
- **Type II error:** We conclude that there is not enough evidence of improvement when the training program actually does improve productivity.

## Question 4
### Explain the difference between a one-tailed and two-tailed test. Give an example of each.

A **one-tailed test** checks for an effect in **one specific direction**.

**Example:**
- H₀: μ = 10
- H₁: μ > 10

This is a **right-tailed test** because we are testing whether the mean is greater than 10.

A **two-tailed test** checks for a difference in **either direction**.

**Example:**
- H₀: μ = 10
- H₁: μ ≠ 10

This tests whether the mean is either significantly greater than or significantly less than 10.

## Question 5
### A company claims that the average time to resolve a customer complaint is 10 minutes.

A random sample of **9 complaints** has:

- Sample mean = **12 minutes**
- Sample standard deviation = **3 minutes**
- α = **0.05**

Because the population standard deviation is not given and the sample size is small, we use a **one-sample t-test**.

#### Hypotheses

- H₀: μ = 10 minutes
- H₁: μ ≠ 10 minutes

This is a **two-tailed test**.

```python
import scipy.stats as stats
import math

# Question 5 data
n = 9
sample_mean = 12
sample_sd = 3
mu0 = 10
alpha = 0.05

t_stat = (sample_mean - mu0) / (sample_sd / math.sqrt(n))
df = n - 1
p_value = 2 * stats.t.sf(abs(t_stat), df)

critical_t = stats.t.ppf(1 - alpha/2, df)

print(f"t-statistic = {t_stat:.4f}")
print(f"Degrees of freedom = {df}")
print(f"Two-tailed p-value = {p_value:.4f}")
print(f"Critical values = ±{critical_t:.4f}")

if p_value <= alpha:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

### Conclusion for Question 5

The calculated t-statistic is **2.00**, with **8 degrees of freedom**. The two-tailed p-value is approximately **0.0805**.

Since **0.0805 > 0.05**, we **fail to reject H₀**.

**Final conclusion:** At the 5% significance level, there is **not enough evidence to conclude that the average complaint-resolution time differs from 10 minutes**. Therefore, the company's claim is **not rejected**.

## Question 6
### When should you use a Z-test instead of a t-test?

A **Z-test** is generally used when:

1. The population standard deviation (σ) is known, and
2. The population is normally distributed or the sample size is sufficiently large.

A **t-test** is preferred when the population standard deviation is **unknown**, especially for small samples.

**In short:**

- **Z-test:** population σ known (or large-sample conditions justify the approximation)
- **t-test:** population σ unknown, particularly with a small sample

## Question 7
### Productivity of 6 employees before and after training

The assignment gives the following paired observations:

| Employee | Before | After |
|---:|---:|---:|
| 1 | 50 | 55 |
| 2 | 60 | 65 |
| 3 | 58 | 59 |
| 4 | 55 | 58 |
| 5 | 62 | 63 |
| 6 | 56 | 59 |

At **α = 0.05**, test whether the training **improved productivity**.

Because the same employees were measured before and after training, we use a **paired t-test**.

#### Hypotheses

Let d = After − Before.

- H₀: μd = 0 (no improvement)
- H₁: μd > 0 (training improved productivity)

This is a **one-tailed paired t-test**.

```python
import numpy as np
from scipy.stats import ttest_rel, t

before = np.array([50, 60, 58, 55, 62, 56])
after  = np.array([55, 65, 59, 58, 63, 59])

differences = after - before
n = len(differences)
mean_diff = differences.mean()
sd_diff = differences.std(ddof=1)

t_stat = mean_diff / (sd_diff / math.sqrt(n))
df = n - 1

# One-tailed p-value for H1: mean difference > 0
p_value = t.sf(t_stat, df)
critical_t = t.ppf(1 - 0.05, df)

print("Differences (After - Before):", differences)
print(f"Mean difference = {mean_diff:.4f}")
print(f"Standard deviation of differences = {sd_diff:.4f}")
print(f"t-statistic = {t_stat:.4f}")
print(f"Degrees of freedom = {df}")
print(f"One-tailed p-value = {p_value:.6f}")
print(f"Critical t-value = {critical_t:.4f}")

if p_value <= 0.05:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

### Conclusion for Question 7

The mean improvement in productivity is **3 units**. The calculated t-statistic is approximately **5.12**, with **5 degrees of freedom**, and the one-tailed p-value is approximately **0.0019**.

Since **p-value < 0.05**, we **reject H₀**.

**Final conclusion:** There is sufficient statistical evidence at the 5% significance level to conclude that the training program **improved employee productivity**.

## Question 8
### Test whether product preference is independent of gender

The assignment provides this contingency table:

| Gender | Product A | Product B | Total |
|---|---:|---:|---:|
| Male | 30 | 20 | 50 |
| Female | 10 | 40 | 50 |
| **Total** | **40** | **60** | **100** |

At **α = 0.05**, test whether product preference is independent of gender.

The appropriate test is a **Chi-square test of independence**.

#### Hypotheses

- H₀: Product preference is independent of gender.
- H₁: Product preference is not independent of gender.

```python
import numpy as np
from scipy.stats import chi2_contingency

observed = np.array([
    [30, 20],
    [10, 40]
])

chi2_stat, p_value, dof, expected = chi2_contingency(observed, correction=False)

print("Observed frequencies:")
print(observed)

print("\nExpected frequencies:")
print(expected)

print(f"\nChi-square statistic = {chi2_stat:.4f}")
print(f"Degrees of freedom = {dof}")
print(f"p-value = {p_value:.6f}")

if p_value <= 0.05:
    print("Decision: Reject H0")
else:
    print("Decision: Fail to reject H0")
```

### Conclusion for Question 8

The expected frequencies are:

- Male, Product A = 20
- Male, Product B = 30
- Female, Product A = 20
- Female, Product B = 30

The calculated **χ² statistic = 16.67**, with **1 degree of freedom**, and the p-value is approximately **0.000045**.

Since **p-value < 0.05**, we **reject H₀**.

**Final conclusion:** There is sufficient evidence to conclude that **product preference and gender are not independent**. In other words, product preference is significantly associated with gender.
